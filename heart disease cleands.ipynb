```python
import pandas as pd
```

```python
pd.read_excel("heart-2.xlsx")
df=pd.read_excel("heart-2.xlsx")
df.head()
```

```python
# Shape of dataset
print("Dataset Shape:", df.shape)
```

```python
# Dataset information
print(df.info())
```

```python
#Summary Statistics
df.describe()
```

```python
# 5. Check Missing Values

missing_values = df.isna().sum()

print("Missing Values:\n")
print(missing_values)
```

```python
#Check Duplicate Rows

duplicate_rows = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)
```

```python
# 7. Remove Duplicate Rows

df.drop_duplicates(inplace=True)

print("New dataset shape after removing duplicates:")
print(df.shape)
```

```python
# Sex mapping
df['sex'] = df['sex'].map({
    0: 'female',
    1: 'male'
})
```

```python
# Chest Pain Type
df['cp'] = df['cp'].map({
    0: 'typical_angina',
    1: 'atypical_angina',
    2: 'non_anginal_pain',
    3: 'asymptomatic'
})
```

```python
# Fasting Blood Sugar
df['fbs'] = df['fbs'].map({
    0: 'normal',
    1: 'high'
})
```

```python
# Exercise Induced Angina
df['exang'] = df['exang'].map({
    0: 'no',
    1: 'yes'
})
```

```python
# Heart Disease Target
df['target'] = df['target'].map({
    0: 'disease',
    1: 'no_disease'
})

df.head()
```

```python
print("Minimum Blood Pressure:", df['trestbps'].min())
print("Minimum Cholesterol:", df['chol'].min())
print("Minimum Heart Rate:", df['thalach'].min())
```

```python
 #Separate Features and Target

target_col = 'target'

X = df.drop(columns=[target_col])

y = df[target_col]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
#Save Cleaned Dataset


output_path = 'heart_cleaned.csv'

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
```
